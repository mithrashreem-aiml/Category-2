```python
# Fully Observable AI Agent
# Smart Traffic Signal Control Agent

class TrafficSignalAgent:

    def __init__(self):
        self.state = {}

    # Agent receives complete environment information
    def perceive(self, traffic_data):
        self.state = traffic_data

    # Decision making
    def decide_action(self):

        north = self.state["North"]
        south = self.state["South"]
        east = self.state["East"]
        west = self.state["West"]
        emergency = self.state["Emergency"]

        if emergency:
            return "Give Green Signal to Emergency Vehicle Lane"

        max_traffic = max(
            north,
            south,
            east,
            west
        )

        if max_traffic == north:
            return "Green Signal: North Road"

        elif max_traffic == south:
            return "Green Signal: South Road"

        elif max_traffic == east:
            return "Green Signal: East Road"

        else:
            return "Green Signal: West Road"


# Environment State (Fully Observable)
environment = {
    "North": 40,
    "South": 25,
    "East": 70,
    "West": 30,
    "Emergency": False
}


agent = TrafficSignalAgent()

agent.perceive(environment)

print("Current Environment:")
print(environment)

print("\nAgent Decision:")
print(agent.decide_action())
```

```python
# Partially Observable AI Agent

class MedicalAgent:

    def __init__(self):
        self.belief = {}

    def perceive(self, symptoms):
        self.belief.update(symptoms)

    def decide(self):

        if self.belief.get("fever") and self.belief.get("cough"):
            return "Possible Infection"

        elif self.belief.get("headache"):
            return "Suggest Rest"

        else:
            return "Need More Information"


agent = MedicalAgent()

observation = {
    "fever": True,
    "cough": True
}

agent.perceive(observation)

print("Agent Decision:", agent.decide())
```

```python
# Static Environment Agent

class ChessAgent:

    def evaluate(self, board):
        if board == "Winning Position":
            return "Make Best Move"
        else:
            return "Analyze Board"


agent = ChessAgent()

board_state = "Winning Position"

print(agent.evaluate(board_state))
```

```python
# Dynamic Environment Agent

class CarAgent:

    def perceive(self, traffic, obstacle):
        self.traffic = traffic
        self.obstacle = obstacle

    def action(self):

        if self.obstacle:
            return "Apply Brake"

        elif self.traffic == "Heavy":
            return "Reduce Speed"

        else:
            return "Continue Driving"


car = CarAgent()

car.perceive("Heavy", False)

print("Action:", car.action())
```

```python
# Discrete Environment Agent

class GameAgent:

    def move(self, state):

        actions = {
            "Start": "Move Right",
            "Enemy": "Attack",
            "Goal": "Stop"
        }

        return actions.get(state)


agent = GameAgent()

state = "Enemy"

print("Agent Action:", agent.move(state))
```

```python
# Continuous Environment Agent

class TemperatureAgent:

    def decide(self, temperature):

        if temperature > 30:
            return "Turn ON AC"

        elif temperature < 18:
            return "Turn ON Heater"

        else:
            return "Maintain Temperature"


agent = TemperatureAgent()

current_temperature = 35

print(agent.decide(current_temperature))
```

```python
# Single Agent Environment
# Smart Farming Agent

class FarmingAgent:

    def perceive(self, soil, temperature, humidity):
        self.soil = soil
        self.temperature = temperature
        self.humidity = humidity

    def decide(self):

        if self.soil < 30:
            return "Activate Irrigation System"

        elif self.temperature > 35:
            return "Provide Cooling Measures"

        elif self.humidity < 40:
            return "Increase Water Supply"

        else:
            return "Crop Conditions are Optimal"


agent = FarmingAgent()

agent.perceive(
    soil=20,
    temperature=30,
    humidity=50
)

print(agent.decide())
```

```python
# Multi Agent Environment
# Smart Warehouse Robot System

class WarehouseRobot:

    def __init__(self, name):
        self.name = name

    def action(self, task):
        if task == "Pick":
            return self.name + " picks the product"

        elif task == "Deliver":
            return self.name + " delivers the product"

        else:
            return self.name + " waits"


# Creating multiple agents

robot1 = WarehouseRobot("Robot A")
robot2 = WarehouseRobot("Robot B")
robot3 = WarehouseRobot("Robot C")


print(robot1.action("Pick"))
print(robot2.action("Deliver"))
print(robot3.action("Wait"))
```

```python
# Deterministic Environment Agent
# Chess Move Agent

class ChessAgent:

    def perceive(self, board):
        self.board = board

    def decide(self):

        if self.board == "Opponent King Exposed":
            return "Move Queen for Checkmate"

        elif self.board == "Opponent Piece Nearby":
            return "Capture Piece"

        else:
            return "Develop Pieces"


agent = ChessAgent()

environment = "Opponent King Exposed"

agent.perceive(environment)

print("Agent Action:", agent.decide())
```

```python
# Stochastic Environment Agent
# Stock Market Trading Agent

import random


class TradingAgent:

    def __init__(self):
        self.balance = 10000
        self.stock = 0

    def perceive(self, price, trend):
        self.price = price
        self.trend = trend

    def decide_action(self):

        if self.trend == "Increasing":
            return "BUY"

        elif self.trend == "Decreasing":
            return "SELL"

        else:
            return "HOLD"


    def market_result(self, action):

        outcomes = {
            "BUY": ["Profit", "Loss"],
            "SELL": ["Profit", "Loss"],
            "HOLD": ["No Change"]
        }

        return random.choice(outcomes[action])


agent = TradingAgent()


# Current market state

price = 150
trend = random.choice(
    ["Increasing", "Decreasing", "Stable"]
)


agent.perceive(price, trend)


action = agent.decide_action()

result = agent.market_result(action)


print("Stock Price:", price)
print("Market Trend:", trend)
print("Agent Action:", action)
print("Market Result:", result)
```

```python
#Episodic Environment (Fruit Inspector)
import random

class FruitInspector:
    def inspect(self, color, size):
        if color == "red" and size == "large":
            return "Premium"
        elif color == "red":
            return "Standard"
        else:
            return "Reject"

agent = FruitInspector()

print(type(agent))   # Type command

colors = ["red", "green", "yellow"]
sizes = ["small", "medium", "large"]

for i in range(5):
    color = random.choice(colors)
    size = random.choice(sizes)
    print(f"Fruit {i+1}: {color}, {size} -> {agent.inspect(color, size)}")
```

```python
#Sequential Environment (Warehouse Robot)
class DeliveryRobot:

    def __init__(self):
        self.position = 0

    def move(self, command):
        if command == "RIGHT":
            self.position += 1
        elif command == "LEFT":
            self.position -= 1

        print("Current Position:", self.position)

robot = DeliveryRobot()



commands = ["RIGHT", "RIGHT", "LEFT", "RIGHT"]

for c in commands:
    robot.move(c)
```

```python
#Known Environment (Parking Fee)
class ParkingFeeAgent:

    def calculate_fee(self, hours):
        if hours <= 2:
            return 20
        elif hours <= 5:
            return 50
        else:
            return 100

agent = ParkingFeeAgent()



hours = [1, 3, 6]

for h in hours:
    print("Hours:", h, "Fee:", agent.calculate_fee(h))
```

```python
#Unknown Environment (Plant Disease)
class PlantDiseaseAgent:

    def __init__(self):
        self.knowledge = {}

    def diagnose(self, symptom):
        if symptom in self.knowledge:
            print("Disease:", self.knowledge[symptom])
        else:
            disease = input("Unknown symptom. Enter disease: ")
            self.knowledge[symptom] = disease
            print("Knowledge Updated")

agent = PlantDiseaseAgent()

print(type(agent))   # Type command

while True:
    symptom = input("Enter symptom (exit to stop): ")
    if symptom.lower() == "exit":
        break
    agent.diagnose(symptom)
```
